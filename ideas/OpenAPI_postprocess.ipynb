```python
import os
import numpy as np
import pandas as pd

date = 20220822

for dmobscd in os.listdir('../data/OpenAPI/dm/'):
    df_dm = []
    for file_name in sorted(os.listdir(f'../data/OpenAPI/dm/{dmobscd}')):
        file_path = f'../data/OpenAPI/dm/{dmobscd}/{file_name}'
        df_dm.append( pd.read_csv(file_path, parse_dates=['ymdhm']).iloc[::-1] )
    df_dm = pd.concat(df_dm, axis=0, ignore_index=True)
    df_dm.to_csv(f'../data/OpenAPI_merged/dm_{date}.csv', index=False)
```
